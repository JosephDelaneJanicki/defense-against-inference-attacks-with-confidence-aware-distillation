```python
import numpy as np
import matplotlib.pyplot as plt

train_conf = np.load("../results/train_confidence.npy")
test_conf = np.load("../results/test_confidence.npy")

plt.hist(train_conf, bins=50, alpha=0.5, label="Train (Members)")
plt.hist(test_conf, bins=50, alpha=0.5, label="Test (Non-Members)")
plt.legend()
plt.title("Confidence Distribution: Train vs Test")
plt.xlabel("Confidence")
plt.ylabel("Frequency")
plt.show()
```
